```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel("../Data/supermarket_sales_cleaned.xlsx")
```

```python
df.shape
```

```python
product_sales = df.groupby("Product line")["Sales"].sum()

product_sales
```

```python
product_sales.plot(kind="bar")

plt.title("Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
City_sales = df.groupby("City")["Sales"].sum()
ax = City_sales.plot(kind="bar")

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

for i, value in enumerate(City_sales):
    ax.text(i, value, round(value), ha="center")

plt.show()
```

```python
Gender_sales = df.groupby("Gender")["Sales"].sum()
ax=Gender_sales.plot(kind="bar")

plt.title("Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
for i, value in enumerate(Gender_sales):
    ax.text(i, value, round(value), ha="center")
plt.show()
```

```python
customer_sales = df.groupby("Customer type")["Sales"].sum()
customer_sales.plot(kind="pie", autopct="%1.1f%%")

plt.title("Sales by Customer Type")
plt.ylabel("")
plt.show()
```

```python
payment_counts = df["Payment"].value_counts()
payment_counts.plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Method Usage")
plt.ylabel("")
plt.show()
```

```python
payment_counts = df["Payment"].value_counts()
payment_counts
```

```python
payment_counts.plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Method Usage")
plt.ylabel("")

plt.show()
```

```python
monthly_sales = df.groupby(df["Date"].dt.month_name())["Sales"].sum()

monthly_sales
```

```python
ax = monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
for i, value in enumerate(monthly_sales):
    ax.text(i, value,round(value), ha="center")
plt.ylabel("Total Sales")
plt.show()
```

```python
product_rating = df.groupby("Product line")["Rating"].mean()
product_rating
```

```python
ax = product_rating.plot(kind="barh")

plt.title("Average Rating by Product Line")
plt.xlabel("Average Rating")
plt.ylabel("Product Line")

for i, value in enumerate(product_rating):
    ax.text(value, i, round(value, 2), va="center")

plt.show()
```

## Business Insights

* Total sales were approximately ₹322,966.75 across 1,000 transactions.
* Food and beverages generated the highest sales among all product lines.
* Giza (Naypyitaw) recorded the highest total sales among the three cities.
* Female customers contributed higher total sales than male customers.
* Member customers generated higher total sales than normal customers.
* E-wallet was the most frequently used payment method.
* Food and beverages had the highest average customer rating at approximately 7.11.
* Home and lifestyle had the lowest average customer rating at approximately 6.84.
* The highest single transaction was ₹1,042.65.
